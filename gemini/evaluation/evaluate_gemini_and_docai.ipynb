## Define custom metrics for JSON schema validation

````python
# Check that the reference value is valid according to the schema
validate(instance=json.loads(reference), schema=schema)
````

````python
def is_valid_schema(instance: dict[str,str]) -> dict[str, float]:
    """Return 1 if the response complies with the schema, 0 if not"""
    
    response = instance["response"]
    
    try:
        validate(instance=json.loads(response), schema=schema)
    except:
        return {
            "valid_schema": 0
        }
    
    return {
        "valid_schema": 1
    }
            
valid_schema = CustomMetric(name="valid_schema", metric_function=is_valid_schema)
````

````python
def calculate_accuracy(instance: dict[str,str]) -> dict[str, float]:
    """
    This calculates accuracy as equality between the JSON strings by 
    converting to DataFrames and using DataFrame.eq to test of equality elementwise.
    
    Overall accuaracy is calculated between 0 (no elements are equal) and 1.0 (all elements are equal).
    """
    
    ref_rows = json.loads(instance["reference"])["rows"]
    resp_rows = json.loads(instance["response"])["rows"]
    
    # Convert reference and response to DataFrame
    reference_df = pd.DataFrame(ref_rows, index=pd.RangeIndex(len(ref_rows)))
    response_df = pd.DataFrame(resp_rows, index=pd.RangeIndex(len(resp_rows)))
    
    print(f"response: {response_df}")
    
    comparison = reference_df.eq(response_df)
    
    return {
        "accuracy": comparison.mean(axis=1)[0]
    }

accuracy = CustomMetric(name="accuracy", metric_function=calculate_accuracy)   
````
